```python
from dataclasses import dataclass

@dataclass
class JudgeConfig:
    pass_score: int = 80
    conditional_score: int = 60
```

```python
async def fake_judge(
    prompt: str,
    config: JudgeConfig,
) -> str:
    return """
    {
        "score": 88,
        "verdict": "PASS",
        "reason": "불만 원인과 개선안이 연결되고 실행 가능함"
    }
    """
```

```python
config = JudgeConfig(
pass_score=80,
conditional_score=60,
)

result = await fake_judge(
prompt="배송 지연 고객불만에 대한 개선안을 평가하세요.",
config=config,
)

result
```
